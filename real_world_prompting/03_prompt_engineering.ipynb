# 第三课：提示词工程

在第一课中，我们快速回顾了一些关键的提示技巧。在第二课中，我们编写了一个"盲目"应用所有这些技巧的提示词。理解这些技巧至关重要，但同样重要的是理解提示词工程的工作流程和决策框架。

---

## 什么是提示词工程？

提示词工程是为您的大型语言模型（如 Claude）制作有效指令的艺术和科学，以产生期望的输出。提示词工程的核心涉及设计、改进和优化给予模型的文本输入（提示词），以获取准确、相关且有用的响应。这关乎于以最大化模型对特定任务的理解和性能的方式与 Claude"沟通"。提示词工程的重要性不容低估：

* 增强 AI 能力：精心设计的提示词可以显著提高 AI 的性能，使其能够更准确、更高效地处理复杂任务。
* 弥合人类意图与 AI 输出之间的鸿沟：提示词工程有助于将人类目标转化为 AI 模型能够有效理解和行动的语言。
* 优化资源使用：熟练的提示词工程可以减少 token 使用量，降低生产成本环境中的成本并缩短响应时间。

### 它与"基本提示"有何不同？

让我们将"基本提示"定义为简单地询问 AI 模型一个问题或给出简单指令。另一方面，提示词工程是一个更复杂和深思熟虑的过程。

* 复杂性：基本提示通常涉及单一交互的简单查询。另一方面，提示词工程可能涉及多轮对话、复杂指令以及精心设计的输入和输出结构。
* 精确性：基本提示可能模糊或含糊，导致结果不一致。这在一次性提示情况下可能没问题，但不会扩展到生产用例。工程化的提示是精确的，几乎不给模型误解的空间。
* 迭代改进：与可能是单次活动的基本提示不同，提示词工程涉及随时间对提示词进行系统测试、分析和改进。
* 可扩展性：提示词工程旨在创建能够处理各种输入和用例的提示词，使其适合生产环境。

从本质上讲，提示词工程将与模型的交互从随意对话提升为精心策划的交流，旨在最大化模型在解决现实世界问题中的潜力**可重复地**。随着我们继续本课程的学习，您将学习掌握这项关键技能所需的技术和思维方式。

## 提示词工程生命周期
最好能坐在空白页前一次就能写出完美的提示词，但现实是提示词工程是一个迭代过程，涉及创建、测试和改进提示词以达到最佳性能。


理解这个生命周期对于开发有效提示词和排查出现的问题至关重要。

1. 初始提示词创建
2. 测试和识别问题
3. 选择适当的技术
4. 实施改进
5. 迭代和改进



让我们详细讨论这些步骤。

### 初始提示词创建

首先编写一个"初稿"提示词，清楚表达您的最终目标。理想情况下，这第一次尝试应该包含我们之前介绍的一些提示技巧：定义目标、识别关键信息、良好地组织提示词结构等。无论如何，提示词的最终版本不太可能与这个初稿相似。

### 测试和识别问题

现在您有了一个基本提示词，下一步是针对各种可能的输入进行测试。我们即将推出的提示词评估课程将详细涵盖这个过程，但简而言之：
* 准备测试用例：创建一组多样化的输入，涵盖各种场景和边缘情况。
* 运行初步测试：使用您的提示词和准备好的输入，观察输出。
* 分析结果：对模型的响应进行"评分"。有多种方式进行：通过代码、人工专业知识或使用大型语言模型。

在整个测试过程中，需要牢记的关键指标包括：
* 准确性：输出是否正确且相关？
* 一致性：模型在不同输入下表现是否相似？它如何处理边缘情况？
* 完整性：输出中是否包含所有必需信息？
* 遵循指令：模型是否遵循所有给定指示？


### 选择适当的技术
一旦您测试了初始提示词并识别出具体问题，就该对提示词进行一些更改了。与其盲目进行改进，理想方法涉及以下步骤：

* 诊断根本原因：对于每个识别出的问题，尝试理解为什么会出现这个问题。
* 研究解决方案：根据您的诊断，探索可能解决问题的提示词工程技术。
* 选择技术：选择最有前景的策略来实施。


### 实施改进

接下来，在原始提示词中实际实施改进。修改原始提示词以纳入您选择的技术，这些技术针对先前识别的问题。如果您要进行多次更改，考虑一次实施一个，以更好地了解它们的个别影响。

### 迭代和改进
重复上述过程！

* 重新测试：使用最初相同的测试用例运行更新后的提示词。
* 比较结果：分析输出发生了怎样的变化，以及目标问题是否已解决。
* 识别新问题：寻找可能因您的更改而产生的新问题。
* 重复循环：继续这个测试、分析和改进的过程，直到达到满意的性能。
