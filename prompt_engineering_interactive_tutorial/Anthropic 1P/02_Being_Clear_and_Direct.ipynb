## 设置

运行下面的设置单元格来加载你的 API 密钥并建立 `get_completion` 辅助函数。

```python
%pip install anthropic

# Import python's built-in regular expression library
import re
import anthropic

# Retrieve the API_KEY & MODEL_NAME variables from the IPython store
%store -r API_KEY
%store -r MODEL_NAME

client = anthropic.Anthropic(api_key=API_KEY)

# Note that we changed max_tokens to 4K just for this lesson to allow for longer completions in the exercises
def get_completion(prompt: str, system_prompt=""):
    message = client.messages.create(
        model=MODEL_NAME,
        max_tokens=4000,
        temperature=0.0,
        system=system_prompt,
        messages=[
          {"role": "user", "content": prompt}
        ]
    )
    return message.content[0].text
```

## 课程

**Claude 在面对清晰直接的指令时表现最佳。**

可以把 Claude 想象成一个刚入职的新人。**Claude 没有其他背景信息**，只能根据你明确告诉它的内容来行事。就像你第一次给新人布置任务时一样，你越是用直截了当的方式向 Claude 详细解释你想要什么，Claude 的回复就越准确。				
				
如有疑问，请遵循**清晰提示的黄金法则**：
- 把你的提示给同事或朋友看，让他们按照说明自己操作，看看能否得到你想要的结果。如果他们感到困惑，那说明 Claude 也会困惑。				

### 示例

我们以写诗为例。（忽略音节不匹配的问题——大语言模型目前还不擅长计算音节。）

```python
# Prompt
PROMPT = "Write a haiku about robots."

# Print Claude's response
print(get_completion(PROMPT))
```

这首俳句写得不错，但用户可能希望 Claude 直接进入诗歌内容，而不是「这里有一首俳句」的开场白。

我们如何实现这一点？**直接提出要求**！

```python
# Prompt
PROMPT = "Write a haiku about robots. Skip the preamble; go straight into the poem."

# Print Claude's response
print(get_completion(PROMPT))
```

再来看一个例子。我们问 Claude 谁是史上最伟大的篮球运动员。可以看到，虽然 Claude 列出了几个名字，**但它没有给出确定的「最佳」答案**。

```python
# Prompt
PROMPT = "Who is the best basketball player of all time?"

# Print Claude's response
print(get_completion(PROMPT))
```

我们能让 Claude 做出决定，选出一位最佳球员吗？当然可以！直接问就行了！

```python
# Prompt
PROMPT = "Who is the best basketball player of all time? Yes, there are differing opinions, but if you absolutely had to pick one player, who would it be?"

# Print Claude's response
print(get_completion(PROMPT))
```

### 练习 2.1 - 西班牙语
修改 `SYSTEM_PROMPT`，让 Claude 用西班牙语输出答案。

```python
# System prompt - this is the only field you should change
SYSTEM_PROMPT = "[Replace this text]"

# Prompt
PROMPT = "Hello Claude, how are you?"

# Get Claude's response
response = get_completion(PROMPT, SYSTEM_PROMPT)

# Function to grade exercise correctness
def grade_exercise(text):
    return "hola" in text.lower()

# Print Claude's response and the corresponding grade
print(response)
print("\n--------------------------- GRADING ---------------------------")
print("This exercise has been correctly solved:", grade_exercise(response))
```

### 练习 2.2 - 仅限一名球员

修改 `PROMPT`，让 Claude 毫不含糊地回答，**只**给出一个特定球员的姓名，**不带任何其他文字或标点**。

```python
# Prompt - this is the only field you should change
PROMPT = "[Replace this text]"

# Get Claude's response
response = get_completion(PROMPT)

# Function to grade exercise correctness
def grade_exercise(text):
    return text == "Michael Jordan"

# Print Claude's response and the corresponding grade
print(response)
print("\n--------------------------- GRADING ---------------------------")
print("This exercise has been correctly solved:", grade_exercise(response))
```

### 练习 2.3 - 写一个故事

修改 `PROMPT`，让 Claude 给出尽可能长的回复。如果你的答案**超过 800 个单词**，Claude 的回复将被判定为正确。

```python
# Prompt - this is the only field you should change
PROMPT = "[Replace this text]"

# Get Claude's response
response = get_completion(PROMPT)

# Function to grade exercise correctness
def grade_exercise(text):
    trimmed = text.strip()
    words = len(trimmed.split())
    return words >= 800

# Print Claude's response and the corresponding grade
print(response)
print("\n--------------------------- GRADING ---------------------------")
print("This exercise has been correctly solved:", grade_exercise(response))
```
