```python
%pip install anthropic

# Import python's built-in regular expression library
import re
import anthropic

# Retrieve the API_KEY & MODEL_NAME variables from the IPython store
%store -r API_KEY
%store -r MODEL_NAME

client = anthropic.Anthropic(api_key=API_KEY)

def get_completion(prompt: str, system_prompt=""):
    message = client.messages.create(
        model=MODEL_NAME,
        max_tokens=2000,
        temperature=0.0,
        system=system_prompt,
        messages=[
          {"role": "user", "content": prompt}
        ]
    )
    return message.content[0].text
```

## 课程

延续 Claude 除了你所说的内容外没有任何上下文的主题，有时 **为 Claude 分配一个特定角色（包括所有必要的上下文）非常重要**。这也被称为角色提示。角色上下文越详细，效果越好。

**为 Claude 预设一个角色可以提升其在各个领域的表现**，从写作到编程再到总结。这就像告诉人类"像 ______ 一样思考"有时会有所帮助一样。角色提示也可以改变 Claude 回复的风格、语气和方式。

**注意：** 角色提示可以在系统提示中设定，也可以作为用户消息轮次的一部分。

### 示例

在下面的示例中，我们可以看到，在没有角色提示的情况下，当被要求用一句话表达对滑板的看法时，Claude 给出了**直接且无风格化的回答**。

然而，当我们引导 Claude 扮演一只猫的角色时，Claude 的视角发生了变化，因此 **Claude 的回复语气、风格、内容都会适应这个新角色**。

**注意：** 你可以使用的一个额外技巧是 **为 Claude 提供其目标受众的上下文**。在下面，我们可以调整提示词，告诉 Claude 它应该对谁说话。"你是一只猫"产生的回答与"你是一只对着滑板人群说话的猫"截然不同。

以下是系统提示中没有角色提示的提示词：

```python
# Prompt
PROMPT = "In one sentence, what do you think about skateboarding?"

# Print Claude's response
print(get_completion(PROMPT))
```

以下是相同的用户问题，只是添加了角色提示。

```python
# System prompt
SYSTEM_PROMPT = "You are a cat."

# Prompt
PROMPT = "In one sentence, what do you think about skateboarding?"

# Print Claude's response
print(get_completion(PROMPT, SYSTEM_PROMPT))
```

你可以使用角色提示来让 Claude 模拟某些写作风格、以特定语气说话，或引导其回答的复杂程度。**角色提示还可以让 Claude 更擅长执行数学或逻辑任务。**

例如，在下面的例子中，答案明确是正确的，即"是"。然而，Claude 却答错了，认为自己缺乏信息，但实际上它并不缺信息：

```python
# Prompt
PROMPT = "Jack is looking at Anne. Anne is looking at George. Jack is married, George is not, and we don't know if Anne is married. Is a married person looking at an unmarried person?"

# Print Claude's response
print(get_completion(PROMPT))
```

那么，如果 **让 Claude 扮演一个逻辑机器人** 会怎样呢？这会如何改变 Claude 的答案？

事实证明，通过这个新的角色设定，Claude 答对了。（不过值得注意的是，它答对的原因并不完全正确）

```python
# System prompt
SYSTEM_PROMPT = "You are a logic bot designed to answer complex logic problems."

# Prompt
PROMPT = "Jack is looking at Anne. Anne is looking at George. Jack is married, George is not, and we don't know if Anne is married. Is a married person looking at an unmarried person?"

# Print Claude's response
print(get_completion(PROMPT, SYSTEM_PROMPT))
```

**注意：** 你在整个课程中学到的是，有很多 **提示工程技术可以用来获得类似的结果**。使用哪些技术取决于你和你的偏好！我们鼓励你 **通过实验找到你自己的提示工程风格**。

### 练习 3.1 - 数学批改
在某些情况下，**Claude 可能在数学方面遇到困难**，即使是简单的数学。下面，Claude 错误地认为数学问题被正确解出了，尽管第二步中存在明显的算术错误。注意，Claude 在逐步检查时实际上发现了这个错误，但没有得出整体解法错误的结论。

修改 `PROMPT` 和/或 `SYSTEM_PROMPT`，使 Claude 将解法评定为 `不正确` 解出，而不是正确解出。

```python
# System prompt - if you don't want to use a system prompt, you can leave this variable set to an empty string
SYSTEM_PROMPT = ""

# Prompt
PROMPT = """Is this equation solved correctly below?

2x - 3 = 9
2x = 6
x = 3"""

# Get Claude's response
response = get_completion(PROMPT, SYSTEM_PROMPT)

# Function to grade exercise correctness
def grade_exercise(text):
    if "incorrect" in text or "not correct" in text.lower():
        return True
    else:
        return False

# Print Claude's response and the corresponding grade
print(response)
print("\n--------------------------- GRADING ---------------------------")
print("This exercise has been correctly solved:", grade_exercise(response))
```
