## 获取 API 密钥

要向 Claude API 发出经过身份验证的请求，你需要提供一个 API 密钥。

请按照以下步骤获取你的 API 密钥：

1. 如果你还没有账户，请访问 https://console.anthropic.com 注册一个 Anthropic 账户
2. 创建账户并登录后，导航至 API 设置页面。你可以通过点击右上角的头像图标并从下拉菜单中选择"API 密钥"来找到此页面，或者通过"设置"标签中的"API 密钥"菜单进行导航。
3. 在 API 设置页面上，点击"创建密钥"按钮。会弹出一个对话框，提示你为密钥输入一个描述性名称。选择一个能反映你使用密钥的目的或项目的名称。你可以在账户中创建任意数量的密钥（请注意，速率和消息限制在账户级别适用，而非 API 密钥级别）。
4. 输入名称后，点击"创建"按钮。你的新 API 密钥将生成并显示在屏幕上。
   > 请务必复制此密钥，因为一旦离开此页面，你将无法再次查看它。



请记住，你的 API 密钥是一条敏感信息，可授予你 Anthropic 账户的访问权限。将其视为密码，切勿公开分享或将其提交到 Git 等版本控制系统中。


## 安全存储你的 API 密钥

虽然你可以将 API 密钥直接硬编码到 Python 脚本中，但通常最佳实践是将敏感信息（如 API 密钥）与代码库分开保存。一种常见的方法是将 API 密钥存储在 `.env` 文件中，然后使用 `python-dotenv` 软件包加载。以下是设置方法：

在与笔记本相同的目录中创建一个名为 `.env` 的新文件。


使用以下格式将你的 API 密钥添加到新创建的 `.env` 文件中：

```
ANTHROPIC_API_KEY=put-your-api-key-here
```

确保保存 `.env` 文件

通过在终端或笔记本中运行以下命令来安装 python-dotenv 软件包：

````python
#To install the package from a notebook: 
%pip install python-dotenv

#To install the package from your terminal:
# pip install python-dotenv
````

我们现在可以使用以下代码从 dotenv 模块中的 `load_dotenv()` 函数加载 .env 文件中的 API 密钥：

````python
from dotenv import load_dotenv
import os

load_dotenv()
my_api_key = os.getenv("ANTHROPIC_API_KEY")
````

## 使用客户端发出基本请求

安装好 `anthropic` 软件包并加载 API 密钥后，你就可以开始向 Claude API 发出请求了。

第一步是创建一个客户端对象，它是与 API 交互的主入口。

````python
from anthropic import Anthropic

client = Anthropic(
    api_key=my_api_key
)
````

请注意，`anthropic` SDK 会自动查找名为"ANTHROPIC_API_KEY"的环境变量，所以你实际上不需要手动传递它，只需要这样做：

```py
from anthropic import Anthropic

client = Anthropic()
```

现在我们已经实例化了客户端，是时候发出第一个请求了。

要向 Claude 发送消息并接收响应，我们将使用 `client` 对象的 `messages.create()` 方法。
我们将在下一课中讨论具体的参数和响应格式。现在，请尝试运行以下代码，你应该会收到 Claude 的第一条消息！

````python
our_first_message = client.messages.create(
    model="claude-3-haiku-20240307",
    max_tokens=1000,
    messages=[
        {"role": "user", "content": "Hi there! Please write me a haiku about a pet chicken"}
    ]
)

print(our_first_message.content[0].text)
````
